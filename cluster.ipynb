```python
from __future__ import print_function
%matplotlib inline
import mdtraj as md
import numpy as np
import scipy.cluster.hierarchy

traj1 = md.load("pca_minima1.dcd", top="comp_nowat.prmtop")
```

```python
atom_indices = [a.index for a in traj1.topology.atoms if a.element.symbol != 'H']
distances = np.empty((traj1.n_frames, traj1.n_frames))
for i in range(traj1.n_frames):
    distances[i] = md.rmsd(traj1, traj1, i, atom_indices=atom_indices)
```

```python
beta = 1
index = np.exp(-beta*distances / distances.std()).sum(axis=1).argmax()
print(index)
```

```python
distances = np.empty((traj1.n_frames, traj1.n_frames))
```

```python
for i in range(traj1.n_frames):
    distances[i] = md.rmsd(traj1, traj1, i)
```

```python
distances[i]
```

```python
print('Max pairwise rmsd: %f nm' % np.max(distances))
```

```python
linkage = scipy.cluster.hierarchy.ward(distances)
```

```python
import matplotlib.pyplot as pp
pp.figure()
pp.title('RMSD Ward hierarchical clustering')
graph = scipy.cluster.hierarchy.dendrogram(linkage, no_labels=True, count_sort='descendent')
```
